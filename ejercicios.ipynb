```python
# Indica el número de caracteres de la palabra más larga.

strs = ["flower", "flow", "flight", "flowers"] 

def palabra_mas_larga(strs):

    if not isinstance(strs, list):
        raise ValueError("El parametro debe ser una lista.")
    for palabra in strs:
        if not isinstance(palabra, str):
            raise ValueError("Todos los elementos de la lista deben ser cadenas de texto.")
    if not strs:
        return 0

    return max(len(palabra) for palabra in strs)

print("El número de letras que tiene la palabra más larga son: " + str(palabra_mas_larga(strs))) 
```

```python
#Indicar el numero de caracteres de palabra más larga y mas corta.

strs = ["flower", "flow", "flight", "flowers"] 

def extremos_de_longitud(strs):

    if not isinstance(strs, list):
        raise ValueError("El parametro debe ser una lista.")
    for palabra in strs:
        if not isinstance(palabra, str):
            raise ValueError("Todos los elementos de la lista deben ser cadenas de texto.")
    if not strs:
        return 0

    longitudes = [len(palabra) for palabra in strs]
    return min(longitudes), max(longitudes)

print("La palabra más corta tiene " + str(extremos_de_longitud(strs)[0]) + " letras y la palabra más larga tiene " + str(extremos_de_longitud(strs)[1]) + " letras.") 
```

```python
#prefijo de palabras que se quieren encontrar en la lista de palabras

strs = ["flower", "flow", "flight", "flowers"] 

def prefijo_comun(strs):

    if not isinstance(strs, list):
        raise ValueError("El parametro debe ser una lista.")
    for palabra in strs:
        if not isinstance(palabra, str):
            raise ValueError("Todos los elementos de la lista deben ser cadenas de texto.")
    if not strs:
        return ""

    prefijo = ""
    for i in range(min(len(palabra) for palabra in strs)):
        letra = strs[0][i]
        if all(palabra[i] == letra for palabra in strs):
            prefijo += letra
        else:
            break
    return prefijo

print("El prefijo común de las palabras es: " + prefijo_comun(strs)) 
```
